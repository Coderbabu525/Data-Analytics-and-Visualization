```python
import pandas as pd
import datetime
dataframe = pd.read_csv("Updated Leeds center air quality data 2020.csv", parse_dates=['End Date'])
dataframe.describe()
```

```python
# daily temperature mean
import matplotlib.pyplot as plt
#dataframe['End Date'] =  pd.to_datetime(dataframe['End Date'], format='%d/%m/%Y')
day_wise = dataframe.groupby('End Date').mean()
days = pd.date_range("1/1/2020", periods=366)
day_wise.columns
#column1 = day_wise[['End Date']]
#print(column1)
temp = day_wise[[' M_T']]
plt.figure(figsize=(20,8))
plt.scatter(days, temp, color='green', marker='o')
plt.title("Air Temperature over the year 2020", fontsize = 20)
plt.xlabel("Date", fontsize = 12)
plt.ylabel("Temperature in Celsius", fontsize = 12)
plt.show()
#year_wise = dataframe.groupby(pd.Grouper(key='End Date', freq='Y')).mean()
#year_wise[['CO', ' PM10', ' NOXasNO2', ' O3', ' PM25', ' SO2']]
```

```python
# yearly mean of pollutants
import matplotlib.pyplot as plt
#dataframe['End Date'] =  pd.to_datetime(dataframe['End Date'], format='%d/%m/%Y')
year_wise = dataframe.groupby(pd.Grouper(key='End Date', freq='Y')).mean()
yearly_mean = [year_wise.iloc[0][' PM10'], year_wise.iloc[0][' PM25'], year_wise.iloc[0][' NOXasNO2']]
std_mean = [20, 10, 40]
index = ['PM10', 'PM2.5', 'NO2']

df = pd.DataFrame({'yearly mean': yearly_mean, 'Acceptable mean': std_mean}, index=index)
ax = df.plot.bar(color = ("g", "r"), width= 0.85, rot=90)

# set title
ax.set_title('Acceptable mean & Yearly mean of Pollutants in 2020')

#set the x-axis label
ax.set_xlabel("Pollutants")

#set the y-axis label
ax.set_ylabel("Units in ug/m3")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("AQI data.csv", parse_dates=['date'])
months = pd.date_range("1/2020", periods=12)
month_wise = df.groupby(pd.Grouper(key='date', freq='M')).mean()
plt.figure(figsize=(20,8))
plt.plot(months, month_wise[['AQI']], color='blue', marker='o')
plt.title("Average Air Quality Index per months", fontsize = 20)
plt.xlabel("Date", fontsize = 12)
plt.ylabel("AQI", fontsize = 12)
plt.show()
```
